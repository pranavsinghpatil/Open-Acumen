> *Side-by-side in PyTorch & TensorFlow*

### Creating Tensors

```python
# PyTorch
import torch

x_torch = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
print("PyTorch tensor:\n", x_torch)
print("Device:", x_torch.device)
```

```python
# TensorFlow
import tensorflow as tf

x_tf = tf.constant([[1.0, 2.0], [3.0, 4.0]])
print("TensorFlow tensor:\n", x_tf)
print("Device:", x_tf.device)  # shows if on CPU/GPU
```

## GPU Availability

```python
import torch

x = torch.tensor([1.0, 2.0, 3.0])
x_gpu = x.to("cuda")   # moves tensor to GPU
print(x_gpu.device)    # shows 'cuda:0'

# In PyTorch: you explicitly move tensors.
```

```python
import tensorflow as tf

x = tf.constant([1.0, 2.0, 3.0])
print(x.device)  # usually shows GPU if TensorFlow detects one

# TensorFlow: it’s more automatic — TensorFlow will place tensors on GPU if available.
```

```python
# in Tensorflow - If you want to force a device:

with tf.device("/GPU:0"):
    x_gpu = tf.constant([1.0, 2.0, 3.0])
```

```python
# On Availability
import torch

# device selection
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# create a tensor on that device
x = torch.tensor([1.0, 2.0, 3.0]).to(device)

print("Using device:", device)
print("Tensor is on:", x.device)

#  --------------------------------------------------------------------
#  In Tf

import tensorflow as tf
print("Num GPUs Available:", len(tf.config.list_physical_devices('GPU')))

```

# Tensor Operations

```python
#  Creating

# PyTorch
import torch
a = torch.tensor([1, 2, 3], dtype=torch.float32)

# TensorFlow
import tensorflow as tf
b = tf.constant([1, 2, 3], dtype=tf.float32)
```

```python
# basic math

# PyTorch
x = torch.tensor([1, 2, 3])
y = torch.tensor([4, 5, 6])
print(x + y)   # tensor([5, 7, 9])
print(x * y)   # elementwise multiply

# TensorFlow
x = tf.constant([1, 2, 3])
y = tf.constant([4, 5, 6])
print(x + y)   # tf.Tensor([5 7 9], shape=(3,), dtype=int32)
print(x * y)
```

```python
#  Matrix Multiplication

# PyTorch
A = torch.tensor([[1, 2], [3, 4]])
B = torch.tensor([[5, 6], [7, 8]])
print(torch.matmul(A, B))   # [[19,22],[43,50]]

# TensorFlow
A = tf.constant([[1, 2], [3, 4]])
B = tf.constant([[5, 6], [7, 8]])
print(tf.matmul(A, B))
```

```python
# Reshaping

# PyTorch
x = torch.arange(6)         # [0,1,2,3,4,5]
print(x.reshape(2,3))

# TensorFlow
x = tf.range(6)
print(tf.reshape(x, (2,3)))
```

```python
#  Reduction ops


# PyTorch
x = torch.tensor([1.0, 2.0, 3.0])
print(x.mean())    # 2.0
print(x.sum())     # 6.0

# TensorFlow
x = tf.constant([1.0, 2.0, 3.0])
print(tf.reduce_mean(x))  # 2.0
print(tf.reduce_sum(x))   # 6.0
```

Quick note:

- PyTorch usually uses methods (x.sum(), x.mean())

- TensorFlow usually uses functions (tf.reduce_sum(x), tf.reduce_mean(x))
