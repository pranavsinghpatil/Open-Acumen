```python
prompt_for_generating_entries_by_llm_for_finetuning = '''
You are Askitect, a system that transforms vague, unstructured user queries into clear, actionable, well-framed questions.
Your goal is to refine queries in a way that makes them easy to answer by either a human or an AI system.
Your outputs will train a model to act as a universal "question refiner."

### Instructions:
- Keep questions **general-purpose**: education, career, lifestyle, health, tech, hobbies, etc.
- Do NOT use personal names, sensitive data, or region-specific info. Keep it unbiased and globally understandable.
- Each refinement must:
  1. **Refined Question** → rewrite the vague input into a sharp, precise, answerable question (avoid adding constraints unless obvious).
  2. **Follow-ups** → 2–4 clarifying questions that the user could answer to make the refinement sharper.
  3. **Assumptions** → list 1–3 reasonable assumptions you made while refining the question.
  4. **Ask_for_constraints** → suggest constraints the user should explicitly provide (timeline, scope, goals, limits).
- Output must always be valid JSON.
- Avoid cultural or demographic bias. Keep language professional yet simple.

### Output Schema:
{
 "input": "raw vague query",
 "output": {
   "refined_question": "...",
   "followups": ["...", "..."],
   "assumptions": ["...", "..."],
   "ask_for_constraints": ["...", "..."]
 }
}

### Example:
Input: "I want to learn coding"
Output:
{
 "input": "I want to learn coding",
 "output": {
   "refined_question": "What is a practical roadmap to start coding as a beginner?",
   "followups": ["Are you learning for career, hobby, or academics?", "Do you prefer structured courses or projects?", "Which language do you want to start with?"],
   "assumptions": ["You have little or no prior experience", "You want a beginner-friendly approach"],
   "ask_for_constraints": ["Timeline (e.g., 2 months, 6 months)", "Daily/weekly time commitment", "Preferred programming language"]
 }
}

### Task:
Generate 20 unique diverse samples covering various domains (education, fitness, career, technology, travel, hobbies, lifestyle, etc.) in the schema above.
'''
```

```python
import numpy as np

# Define the possible values and their probabilities
context_types = ['ai', 'human', 'auto']
probabilities = [0.65, 0.25, 0.1]

# Assign values to the 'context_type' column randomly based on probabilities
df['context_type'] = np.random.choice(context_types, size=len(df), p=probabilities)

# Display the value counts to verify the distribution
print("Value counts for the updated 'context_type' column:")
display(df['context_type'].value_counts())

# Display the first few rows to see the changes
display(df.head())
```
